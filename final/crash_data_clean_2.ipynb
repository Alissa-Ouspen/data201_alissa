```python
# Clean "Drivers_License_State"

# Replace "XX" with nan
crash_culled_df["Drivers_License_State"] = crash_culled_df["Drivers_License_State"].replace("XX", np.nan)

# Replace specific Mexican codes with "Mex"
mexican_codes = ["MX-MEX", "MX-ROO", "MX-GRO", "MX-BCN", "MX-CMX"]
crash_culled_df['Drivers_License_State'] = crash_culled_df["Drivers_License_State"].replace(mexican_codes, "Mex")

# Replace Canadian province codes with "Can"
canadian_codes_pattern = r"^(CA-ON|CA-QC|ON|QC|BC|AB|MB|NB|NL|NS|NT|NU|PE|SK|YT)$"
crash_culled_df['Drivers_License_State'] = crash_culled_df["Drivers_License_State"].replace(
    canadian_codes_pattern, "Can", regex=True)

# Standardize US territories to "US-Territory"
us_territories = ["PR", "GU", "VI", "AS", "MP", "UM"]
crash_culled_df["Drivers_License_State"] = crash_culled_df["Drivers_License_State"].replace(us_territories, "US-Territory")

# Standardize other foreign or generic codes to "Foreign/Other"
foreign_other_codes = ["US", "MH", "FM", "NF", "IT"]
crash_culled_df["Drivers_License_State"] = crash_culled_df["Drivers_License_State"].replace(foreign_other_codes, "Foreign/Other")

# print("Unique values after cleaning Drivers_License_State:")
# print(crash_culled_df["Drivers_License_State"].unique())
# print("\nValue counts for Drivers_License_State:")
# print(crash_culled_df["Drivers_License_State"].value_counts(dropna=False))
```

```python
# Clean Intersection_Type (as generated in previous turn)
crash_culled_df["Intersection_Type"] = crash_culled_df["Intersection_Type"].replace(
    ["ROUNDABOUT", "TRAFFIC CIRCLE"], "Roundabout/Traffic Circle"
)

print("Unique values after cleaning Intersection_Type:")
print(crash_culled_df["Intersection_Type"].unique())

# Transform Lane_Type to Parking_Lot_YN (Y/N/nan)
# Create a temporary series for the new values based on Lane_Type
new_parking_lot_yn_values = crash_culled_df["Lane_Type"].astype(str).str.contains("parking lot", case=False, na=False).map({True: "Y", False: "N"})

# Set to NaN where original Lane_Type was NaN
new_parking_lot_yn_values = new_parking_lot_yn_values.mask(crash_culled_df["Lane_Type"].isna(), np.nan)

# Apply the 'check against off road' condition
# If First_Harmful_Event is 'OFF ROAD' (case-insensitive), then it's not a parking lot, set to 'N'
off_road_condition = crash_culled_df["First_Harmful_Event"].astype(str).str.contains("off road", case=False, na=False)
new_parking_lot_yn_values.loc[off_road_condition] = "N"

# Replace the original 'Lane_Type' column with the new values
crash_culled_df["Lane_Type"] = new_parking_lot_yn_values

# Rename the column to reflect its new meaning
crash_culled_df.rename(columns={"Lane_Type": "Parking_Lot_YN"}, inplace=True)

# Display unique values and value counts for the transformed column
print("\nUnique values of transformed \"Parking_Lot_YN\" column:")
print(crash_culled_df["Parking_Lot_YN"].unique())
print("\nValue counts for \"Parking_Lot_YN\" column:")
print(crash_culled_df["Parking_Lot_YN"].value_counts(dropna=False))
```

```python
# Clean Vehicle_Damage: caseblind, "unkonwn" = "other" = nan
# Convert to string and uppercase to handle case-insensitivity
crash_culled_df["Vehicle_Damage"] = crash_culled_df["Vehicle_Damage"].astype(str).str.upper()

# Define values to replace (now all in uppercase)
values_to_replace_vd = ["UNKONWN", "OTHER", "UNKNOWN", "NAN", "NA"]

# Replace these values with np.nan
crash_culled_df["Vehicle_Damage"] = crash_culled_df["Vehicle_Damage"].replace(values_to_replace_vd, np.nan)

print("Unique values after cleaning Vehicle_Damage:")
print(crash_culled_df["Vehicle_Damage"].unique())
print("\nValue counts for Vehicle_Damage:")
print(crash_culled_df["Vehicle_Damage"].value_counts(dropna=False))
```

```python
# Clean Veh_1st_Impact_Loc: "ROOF TOP" = "roof" = "top"; caseblind; OCLOCK = O Clock to O"Clock

# Convert to string and uppercase for case-insensitivity
crash_culled_df["Veh_1st_Impact_Loc"] = crash_culled_df["Veh_1st_Impact_Loc"].astype(str).str.upper()

# Standardize "ROOF TOP", "ROOF", "TOP" to "ROOF TOP"
crash_culled_df["Veh_1st_Impact_Loc"] = crash_culled_df["Veh_1st_Impact_Loc"].replace(
    r"^(ROOF TOP|ROOF|Top)$", "TOP", regex=True
)

# Create a mapping for clock-related terms to standardize them to O"CLOCK format
clock_mapping = {
    "ONE O CLOCK": "ONE O\'CLOCK", "SIX O CLOCK": "SIX O\'CLOCK", "TWELVE O CLOCK": "TWELVE O\'CLOCK",
    "TWO O CLOCK": "TWO O\'CLOCK", "ELEVEN O CLOCK": "ELEVEN O\'CLOCK", "NINE O CLOCK": "NINE O\'CLOCK",
    "EIGHT O CLOCK": "EIGHT O\'CLOCK", "SEVEN O CLOCK": "SEVEN O\'CLOCK", "FIVE O CLOCK": "FIVE O\'CLOCK",
    "TEN O CLOCK": "TEN O\'CLOCK", "THREE O CLOCK": "THREE O\'CLOCK", "FOUR O CLOCK": "FOUR O\'CLOCK",
    "ONE OCLOCK": "ONE O\'CLOCK", "SIX OCLOCK": "SIX O\'CLOCK", "TWELVE OCLOCK": "TWELVE O\'CLOCK",
    "TWO OCLOCK": "TWO O\'CLOCK", "ELEVEN OCLOCK": "ELEVEN O\'CLOCK", "NINE OCLOCK": "NINE O\'CLOCK",
    "EIGHT OCLOCK": "EIGHT O\'CLOCK", "SEVEN OCLOCK": "SEVEN O\'CLOCK", "FIVE OCLOCK": "FIVE O\'CLOCK",
    "TEN OCLOCK": "TEN O\'CLOCK", "THREE OCLOCK": "THREE O\'CLOCK", "FOUR OCLOCK": "FOUR O\'CLOCK",
    "OCLOCK": "O\'CLOCK", "O CLOCK": "O\'CLOCK"
}
crash_culled_df["Veh_1st_Impact_Loc"] = crash_culled_df["Veh_1st_Impact_Loc"].replace(clock_mapping)

# Replace "NAN" string with np.nan
crash_culled_df["Veh_1st_Impact_Loc"] = crash_culled_df["Veh_1st_Impact_Loc"].replace("NAN", np.nan)

print("Unique values after cleaning Veh_1st_Impact_Loc:")
print(crash_culled_df["Veh_1st_Impact_Loc"].unique())
print("\nValue counts for Veh_1st_Impact_Loc:")
print(crash_culled_df["Veh_1st_Impact_Loc"].value_counts(dropna=False))
```

```python
import numpy as np

# Rename the column first
crash_culled_df = crash_culled_df.rename(columns={"Road Alignment": "Road_Alignment"})

# Preserve original NaN values
original_nan_mask = crash_culled_df["Road_Alignment"].isna()

# Convert to string and title case for standardization, but only for non-NaN values
# This will turn np.nan into the string 'Nan'
crash_culled_df["Road_Alignment"] = crash_culled_df["Road_Alignment"].astype(str).str.title()

# Standardize 'Straight' values
straight_patterns = ["Straight", "Straight, Straight"]
crash_culled_df["Road_Alignment"] = crash_culled_df["Road_Alignment"].replace(straight_patterns, "Straight")

# Convert all variations of 'Curve' to 'Curved'
curve_patterns = [
    "Curve Left", "Curve Left, Straight", "Curve Right",
    "Curve Left, Curve Right", "Curve Right, Straight", "Curve Left, Curve Right, Straight"
]
crash_culled_df["Road_Alignment"] = crash_culled_df["Road_Alignment"].replace(curve_patterns, "Curved")

# Now, convert the string 'Nan' (which came from original np.nan) back to actual np.nan
crash_culled_df["Road_Alignment"] = crash_culled_df["Road_Alignment"].replace("Nan", np.nan)

# Ensure that any values that were originally NaN are still NaN after all processing
crash_culled_df.loc[original_nan_mask, "Road_Alignment"] = np.nan

print("Unique values after cleaning Road_Alignment:")
print(crash_culled_df["Road_Alignment"].unique())
print("\nValue counts for Road_Alignment:")
print(crash_culled_df["Road_Alignment"].value_counts(dropna=False))
```

####Note on Data
In trying to understand the meaning of certain fields, I may have found an error in the Incidents data description on  
https://data.montgomerycountymd.gov/Public-Safety/Crash-Reporting-Incidents-Data/bhju-22kf/about_data  
The "Direction" field is described as "Location - Direction from mile point."  
The "Distance" field is described as "Location - Distance from mile point."  

However, the Maryland Automated Crash Reporting System (ACRS) Field Reference Guide published on  
https://www.nhtsa.gov/sites/nhtsa.gov/files/documents/acrsfieldreference.pdf
(2017 Maryland State Police)  
section 2.19 states:  
"Distance
Definition:
The distance from the referenced Intersecting Road to the crash site.
Explanation:
The distance in feet or miles from the Intersecting Road to crash site."

Section 2.21:  
"Distance Direction
Definition:
The compass direction describing the direction going from the primary and intersecting
roads to the crash.
Explanation:
Units are described in compass points, i.e. North, South East and West. If the crash
occurred in the intersection, select the compass point referenced in the Mile Point
Direction box above."
