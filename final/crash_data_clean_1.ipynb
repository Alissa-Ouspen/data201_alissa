```python
merged_cleaned_df = merged_df[column_list_abridged].copy()
merged_cleaned_df.head()
```

```python
# For all cells in merged_cleaned_df:
# if value in cell == "unkonwn" or "UNKNOWN" or "other" or "OTHER"
# or "Not Applicable" or "NOT APPLICABLE", then set value to nan
values_to_replace = ["unkonwn", "UNKNOWN", "Unknown", "other", "OTHER", "Other","Not Applicable", "NOT APPLICABLE", "not applicable", "na", "NA", "N/A", "n/a"]

for col in merged_cleaned_df.columns:
    if merged_cleaned_df[col].dtype == "object": # Only apply to object (string) columns
        merged_cleaned_df[col] = merged_cleaned_df[col].replace(values_to_replace, np.nan)

display(merged_cleaned_df.head())
```

```python
# merge agency names
merged_cleaned_df["Agency_Name"] = merged_cleaned_df["Agency_Name"].replace({
    "MONTGOMERY" : "Montgomery County Police",
    "Rockville Police Departme" : "Rockville Police Dept",
    "ROCKVILLE" : "Rockville Police Dept",
    "Takoma Park Police Depart" : "Takoma Park Police Dept",
    "TAKOMA" : "Takoma Park Police Dept",
    "Gaithersburg Police Depar" : "Gaithersburg Police Dept",
    "GAITHERSBURG" : "Gaithersburg Police Dept",
    "Maryland-National Capital" : "MD-Natl Capital Park Pol",
    "MCPARK" : "MoCo Park Police"
})
display(merged_cleaned_df["Agency_Name"].unique())
```

```python
for col in merged_cleaned_df.select_dtypes(include='object').columns:
    unique_values = merged_cleaned_df[col].dropna().unique()

    replacement_map = {}
    grouped_values = {}

    # Group values by their lowercase form
    for value in unique_values:
        lower_value = value.lower()
        if lower_value not in grouped_values:
            grouped_values[lower_value] = []
        grouped_values[lower_value].append(value)

    # Create replacement map: all-caps -> first non-all-caps variant
    for lower_key, variants in grouped_values.items():
        all_caps_variants = [v for v in variants if v.isupper()]
        non_all_caps_variants = [v for v in variants if not v.isupper()]

        if non_all_caps_variants and all_caps_variants:
            # Choose the first non-all-caps variant as the target
            target_variant = non_all_caps_variants[0]
            for ac_variant in all_caps_variants:
                if ac_variant != target_variant: # Ensure no self-replacement if target is also all-caps (unlikely with this logic)
                    replacement_map[ac_variant] = target_variant

    if replacement_map: # Only apply replace if there are actual replacements to make
        merged_cleaned_df[col] = merged_cleaned_df[col].replace(replacement_map)
        print(f"Standardized capitalization for column: {col}")

# Display unique values for a few columns to verify the cleaning
print("\nUnique values after capitalization standardization:")
display(merged_cleaned_df["Agency_Name"].unique())
display(merged_cleaned_df["ACRS_Report_Type"].unique())
display(merged_cleaned_df["Weather"].unique())
```

####Note on Data
In trying to understand the meaning of certain fields, I may have found an error in the Incidents data description on  
https://data.montgomerycountymd.gov/Public-Safety/Crash-Reporting-Incidents-Data/bhju-22kf/about_data  
The "Direction" field is described as "Location - Direction from mile point."  
The "Distance" field is described as "Location - Distance from mile point."  

However, the Maryland Automated Crash Reporting System (ACRS) Field Reference Guide published on  
https://www.nhtsa.gov/sites/nhtsa.gov/files/documents/acrsfieldreference.pdf
(2017 Maryland State Police)  
section 2.19 states:  
"Distance
Definition:
The distance from the referenced Intersecting Road to the crash site.
Explanation:
The distance in feet or miles from the Intersecting Road to crash site."

Section 2.21:  
"Distance Direction
Definition:
The compass direction describing the direction going from the primary and intersecting
roads to the crash.
Explanation:
Units are described in compass points, i.e. North, South East and West. If the crash
occurred in the intersection, select the compass point referenced in the Mile Point
Direction box above."
