###Introduction


This document covers the Palmer Penguins dataset, a morphological measurement dataset containing physical characteristics of three penguin species from the Palmer Archipelago in Antarctica.

Size measurements for adult foraging penguins near Palmer Station, Antarctica

Data were collected and made available by Dr. Kristen Gorman and the Palmer Station, Antarctica LTER, a member of the Long Term Ecological Research Network.

####Data collection method:
From Adélie and Gentoo penguins, 30 nests were randomly sampled from each species per nesting season.  From the Chinstrap penguins, 15 nests were sampled, due to the smaller size of the breeding population.

Excerpt from Gorman, 2014:

"Each season, study nests, where pairs of adults were present, were individually marked and chosen before the onset of egg laying, and consistently monitored. When study nests were found at the one-egg stage, both adults were captured to obtain blood samples used for molecular sexing and SI analyses, and measurements of structural size and body mass."

The original dataset included more information, such as some DNA markers and the year in which each sample was taken.

####Ethics statement from Dr. Gorman's paper(2014):

"Research was conducted in accordance with an Antarctic Conservation Act permit to WRF (2008-020), in addition to Canadian Committee on Animal Care guidelines (Simon Fraser University, SFU, Animal Care Permit 890B-08 to KBG and TDW)."

Limitations of the data and possible bias:  the researchers focused on nests, once at least one egg was present. Samples were collected from individuals who had reproduced successfully.  Therefore, the model predictions for certain physical characteristics can only be applied to penguins successful in breeding and not to all penguins in general, even within species.  For example, malnutrition is often associated with failure to breed, so penguins excluded from the study may skew towards smaller body mass than those in the breeding population.

Originally published in: Gorman KB, Williams TD, Fraser WR (2014) Ecological Sexual Dimorphism and Environmental Variability within a Community of Antarctic Penguins (Genus Pygoscelis). PLoS ONE 9(3): e90081. doi:10.1371/journal.pone.0090081

###Data overview

```python
import seaborn as sns
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
penguins.info()
```

####Variables:
There are 344 rows and 7 columns in the dataset.

| Name            | Type        | Description                                                          |
|:----------------|:------------|:---------------------------------------------------------------------|
| species         | Categorical | Species of penguin (Adelie, Gentoo, Chinstrap)                     |
| island          | Categorical | Observation location (Torgersen, Biscoe, Dream)                      |
| bill_length_mm  | Quantitative| Length of the penguin's bill in millimeters                        |
| bill_depth_mm   | Quantitative| Depth of the penguin's bill in millimeters                         |
| flipper_length_mm| Quantitative| Length of the penguin's flipper in millimeters                     |
| body_mass_g     | Quantitative| Body mass of the penguin in grams                                  |
| sex             | Categorical | The sex of the penguin (Male or Female)                            |

###Quantitative data overview
- all quantitative data is missing for two of the birds

####Bill length:
- from head to tip
- bill lengths range from 32.1 mm to 59.6 mm
- mean bill length is 43.92 mm

####Bill depth:
- thickness, height, virtical diameter
- bill depths range from 13.1 mm to 21.5 mm
- mean bill depth is 17.15 mm

####Flipper length:
- wings used for swimming
- flipper lengths range from 172 mm to 231 mm
- mean flipper length is 200.92 mm

####Body mass:
- body mass ranges from 2700 mm and 6300 mm
- mean flipper body mass is 4201.75 mm
